```python
import pandas as pd

df = pd.read_csv("../AQI_Data/processed_aqi_data.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])

# drop redundant / low-signal / leakage-prone columns
drop_cols = ["location_lat", "location_lon", "aqi_category", "day", "year"]
df = df.drop(columns=drop_cols)

# duplicate check (same location + time shouldn't repeat)
df = df.drop_duplicates(subset=["timestamp", "location_name"])

# sanity: no negative pollutant/aqi values
num_cols = ["co","no2","o3","pm10","pm25","so2","aqi"]
assert (df[num_cols] < 0).sum().sum() == 0, "Negative values found!"

# sort by location + time (important before any lag/shift feature later)
df = df.sort_values(["location_name","timestamp"]).reset_index(drop=True)

print(df.shape)
print(df.dtypes)

df.to_csv("../AQI_Data/clean_aqi_data.csv", index=False)
```
